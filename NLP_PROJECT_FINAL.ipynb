```python
import docx2txt
from pdfminer.high_level import extract_text

def read_file(filename):
    if filename.endswith('.pdf'):
        text = extract_text(filename)
    elif filename.endswith('.docx'):
        text = docx2txt.process(filename)
    else:
        raise Exception("Unsupported file type")
    return text

filename = list(uploaded.keys())[0]
resume_text = read_file(filename)
```

```python
import spacy
import re

# Load spaCy English model
nlp = spacy.load('en_core_web_sm')
```

```python
def extract_name_from_top(text):
    lines = text.split('\n')
    for line in lines:
        if line.strip() and len(line.strip().split()) <= 4 and not any(word in line.lower() for word in ['linkedin', 'github', '.com', '@']):
            return line.strip()
    return None
```

```python
def extract_email(text):
    match = re.search(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)
    return match.group(0) if match else None
```

```python
# Extract Phone
def extract_phone(text):
    phone = re.findall(r'\+?\d[\d\-\s]{8,}\d', text)
    return phone[0] if phone else None
```

```python
# Extract Skills
def extract_skills(text):
    skills = ['python', 'java', 'c++', 'sql', 'html', 'css', 'javascript',
              'react', 'node.js', 'django', 'flask', 'machine learning',
              'data science', 'cloud', 'aws', 'azure', 'docker', 'kubernetes']
    found_skills = []
    for skill in skills:
        if skill.lower() in text.lower():
            found_skills.append(skill)
    return list(set(found_skills))
```

```python
def extract_education(text):
    education = []
    lines = text.split('\n')
    for line in lines:
        line_clean = line.strip().lower()
        if any(degree in line_clean for degree in ['b.tech', 'b.e', 'b.sc', 'm.tech', 'm.sc', 'ph.d', 'hsc', 'sslc']) \
           or any(keyword in line_clean for keyword in ['college', 'school', 'university', 'institute']) \
           or re.search(r'\d{4}\s*-\s*\d{4}', line_clean) \
           or 'cgpa' in line_clean or '%' in line_clean:
            education.append(line.strip())
        # Stop adding if 'certificate', 'project', 'achievement' comes
        elif any(word in line_clean for word in ['certificate', 'certified', 'certification', 'project', 'achievement', 'award', 'workshop']):
            break
    return education
```

```python
def extract_certifications(text):
    certs = []
    lines = text.split('\n')
    capture = False
    for line in lines:
        line_clean = line.strip().lower()
        if 'certificate' in line_clean or 'certified' in line_clean or 'award' in line_clean or 'badge' in line_clean or 'achievement' in line_clean or 'workshop' in line_clean:
            certs.append(line.strip())
    return certs
```
